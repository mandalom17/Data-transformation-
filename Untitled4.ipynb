Q1. Handle Missing Values (Basic Cleaning)

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("employee_productivity_dataset.csv")

# Fill missing values according to specifications
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df["Hours_Worked_Per_Week"] = df["Hours_Worked_Per_Week"].fillna(
    df["Hours_Worked_Per_Week"].median()
)
df["Performance_Score"] = df["Performance_Score"].fillna(
    df["Performance_Score"].mean()
)

# Display the cleaned dataset info and first few rows
print("Missing values after cleaning:\n", df.isnull().sum())
print("\nFirst 5 rows:")
display(df.head())
```

Q2. Label Encoding

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

# Apply label encoding
df["Gender_Encoded"] = le.fit_transform(df["Gender"])
df["Department_Encoded"] = le.fit_transform(df["Department"])

# Show the updated columns
print(df[["Gender", "Gender_Encoded", "Department", "Department_Encoded"]].head())
```

Q3. One-Hot Encoding

```python
# Apply one-hot encoding
df = pd.get_dummies(df, columns=["Work_Mode", "Location"], prefix=["WM", "Loc"])

# Check columns created
print("Total columns after One-Hot Encoding:", len(df.columns))
print(df.filter(like="WM_").head(3))
print(df.filter(like="Loc_").head(3))
```

Q4. Normalization (Min-Max Scaling)

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[["Salary_Normalized", "Hours_Normalized"]] = scaler.fit_transform(
    df[["Salary", "Hours_Worked_Per_Week"]]
)

print(
    df[
        [
            "Salary",
            "Salary_Normalized",
            "Hours_Worked_Per_Week",
            "Hours_Normalized",
        ]
    ].head()
)
```

Q5. Standardization (Scaling)

```python
from sklearn.preprocessing import StandardScaler

std_scaler = StandardScaler()

df[["Age_Standardized", "Projects_Standardized"]] = std_scaler.fit_transform(
    df[["Age", "Projects_Completed"]]
)

print(
    df[
        [
            "Age",
            "Age_Standardized",
            "Projects_Completed",
            "Projects_Standardized",
        ]
    ].head()
)
```

Q6. Compare Scaling Methods

```python
comparison_df = pd.DataFrame()
comparison_df["Original_Salary"] = df["Salary"]
comparison_df["MinMax_Scaled"] = MinMaxScaler().fit_transform(
    df[["Salary"]]
)
comparison_df["Standard_Scaled"] = StandardScaler().fit_transform(
    df[["Salary"]]
)

print(comparison_df.head(10))
```

Q7. Build Preprocessing Pipeline

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

# Define features
numeric_features = [
    "Age",
    "Hours_Worked_Per_Week",
    "Projects_Completed",
    "Performance_Score",
    "Salary",
]
categorical_features = ["Gender", "Department", "Work_Mode", "Location"]

# Numerical pipeline: Impute with median + StandardScaler
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
    ]
)

# Categorical pipeline: Impute with most frequent + OneHotEncoder
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore")),
    ]
)

# Combine into ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features),
    ]
)

print("Preprocessing pipeline built successfully!")
```

Q8. Apply Pipeline

```python
# Reload raw data to apply pipeline from scratch
df_raw = pd.read_csv("employee_productivity_dataset.csv")

# Apply preprocessor
transformed_data = preprocessor.fit_transform(df_raw)

print("Transformed dataset preview (first 5 rows):")
print(transformed_data[:5])
print("\nShape of final dataset:", transformed_data.shape)
```

Q9. Conceptual Question

Why is scaling important in Python?


Answer:
Equal Feature Weighting: Machine learning algorithms (like Gradient Descent-based models, KNN, SVM, and Neural Networks) use distance metrics or weight gradients.

Features with larger numeric ranges (e.g., Salary in tens of thousands) would disproportionately dominate features with smaller ranges (e.g., Performance_Score between 1 and 5). Scaling ensures all features contribute equally.

Faster Convergence: Optimization algorithms reach the global minimum much faster when features are on the same scale (e.g., centered around 0 with unit variance).

Q10. Conceptual Question

Why do we convert categorical data into numerical form?

Answer:
Machine learning models and mathematical algorithms cannot process string labels (like text categories for Gender, Department, or Location). They require mathematical operations and vectors. Converting them via encoding methods (Label or One-Hot Encoding) allows algorithms to interpret categorical variables correctly as numbers or binary flags.
